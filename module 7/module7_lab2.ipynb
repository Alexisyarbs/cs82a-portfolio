# Problem 7.1: What drives a restaurant’s overall rating?

Fit a linear regression predicting the overall rating from ambience, food, and service ratings. With multiple features, model.coef_ returns one slope per feature: rank them and answer the business question: which factor should a struggling restaurant fix first?

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df=pd.read_csv("ratings.tsv", sep="\t")

df.head()
```

```python
X = df[["ambience", "food", "service"]]
y=df["rating"]

model=LinearRegression()
model.fit(X,y)

coefficients = pd.DataFrame({
    "Factor":X.columns,
    "Impact on Overall Ratings": model.coef_
})

ranked_coefficients = coefficients.sort_values(by="Impact on Overall Ratings")

print("Ranked Coefficients:")
print(ranked_coefficients.to_string(index=False))
```

Based on the regression model, food has the most impact on customer satisfaction, with the highest coeeficient of 0.734.
