```python
import pandas as pd
df=pd.read_csv('chickwts.csv')
```

```python
feed_stats=df.groupby('feed')['weight'].agg(['min','max','mean'])
```

```python
print(feed_stats)
```

```python
chosen_feeds= ['casein', 'sunflower']
filtered_df= df[df['feed'].isin(chosen_feeds)]

def get_trimmed_average(values):
    sorted_vals= values.sort_values()
    trimmed_vals = sorted_vals.iloc[1:-1]
    return trimmed_vals.mean() 
```

```python
result = filtered_df.groupby('feed')['weight'].apply(get_trimmed_average)
print(result)
```

I would choose sunflower as my feed of choice. Sunflower has a larger minimum, average, and maxium weight than all the other feeds. The next best choice would be casein since it also has a larger minimum, average, and maxium weight. Since casesin has a smaller span between minimum and maxiumum weights than sunflower, I wanted to account for potential outliers that could affect the average, so I took out the smallest and biggest weight from both feeds then found the new average. Sunflower still had a larger average than casein, which makes it my choice for feed since it would consistently produce larger chickens on average than casein.
