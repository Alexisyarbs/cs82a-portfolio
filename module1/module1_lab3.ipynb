```python
import pandas as pd
df = pd.read_csv('insurance_ratings.csv')

type_column= df.columns[1]
value_column = df.columns[2]

insurance_stats = df.groupby(type_column)[value_column].agg(['min','max','mean'])
print(insurance_stats)
```
