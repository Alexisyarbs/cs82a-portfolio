```python
def multiply (multicand, multiplier):
    result=0

    for _ in range(abs(multiplier)):
        result+= multicand

    if multiplier < 0:
        return -result

    return result

multicand = 7
multiplier = 6
print(multiply(multicand, multiplier))
multicand = -5
multiplier = 10
print(multiply(multicand, multiplier))
multicand = 8
multiplier = -9
print(multiply(multicand, multiplier))
```

```python
def multiply (multicand, multiplier):
    result=0
    counter=0

    while counter < abs(multiplier):
        result += multicand
        counter +=1

    if multiplier <0:
        return -result

    return result

multicand = 7
multiplier = 6
print(multiply(multicand, multiplier))
multicand = -5
multiplier = 10
print(multiply(multicand, multiplier))
multicand = 8
multiplier = -9
print(multiply(multicand, multiplier))
```

Every time the loop runs, it adds 1 to the counter, and the loop is set to terminate once counter equals multiplier.
