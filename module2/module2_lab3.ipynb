
Temperatures under 15 are cold.
Temperatures between 15 and 25 (including 15 and 25 degrees) are warm.
Temperatures over 25 are hot.


```python
temperatures = [10, 16, 23, 28, 14, 22, 30]

for temp in temperatures:
    if temp < 15:
        category = "Cold"
    elif temp <= 25:
        category = "Warm"
    else:
        category = "Hot"
    print (f"Temperature is {temp} degrees which is {category}")
```
