### Find the median of cover types

```python
##sort cover column
sorted_cover = df["cover"].sort_values()
##grab the middle value
median_cover = sorted_cover.iloc[len(sorted_cover)//2]
print(f"The median cover is {median_cover}")
```

### Mean weight of all books

```python
df["weight"].mean()
```

The mean weight of all books is 706.67.

### Mean weight for each cover

```python
df.groupby("cover")["weight"].mean()
```

The mean weight of paperback books is 628.13, while the hardcover books mean weight is 796.43. This shows the type of cover greatly affects the weight of the book.
