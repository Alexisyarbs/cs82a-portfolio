```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("truck.csv")
```

```python
df.shape, df.head()
```

```python
df.columns = ["list_price", "best_price"]
df.head()
```

### Pearson's r between list price and best price

```python
df.corr()
```

Pearson's r between list price and best price is 0.99854.

```python
df["list_price_scaled"] = df["list_price"]*1000
df["best_price_scaled"] = df["best_price"]*1000
df.plot.scatter(x="list_price_scaled", y="best_price_scaled", grid = True)
plt.xlabel("List Price")
plt.ylabel("Best Price")
```

### Prediction
Looking at the scatter plot, I took a list price of 16,000 dolars and found that the best price should be about 14,000 dollars. There is an extremely high level of confidence in this prediction because the scatter plot shows very little variance from the trend line. This shows there is an extremely strong correlation between list price and best price, and since the point between 16,00 and 14,000 falls almost pefectly on the trend line the prediction should be very accurate.
